```python
from transformers import pipeline

# 1. Initialize the text generation pipeline
generator = pipeline(
    "text-generation",
    model="gpt2",
    return_full_text=False, # Return only the generated text, excluding the prompt
    max_new_tokens=50,       # Limit how many new tokens to generate
    #max_length=50,
    do_sample=True,          # Enable sampling for creative generation
    temperature=0.7          # Control randomness
)

# 2. Define a prompt
prompt = "The most important breakthrough in modern artificial intelligence is"

# 3. Generate and print output
output = generator(prompt)
print("Prompt:", prompt)
print("Generated Completion:\n", output[0]['generated_text'])
```
